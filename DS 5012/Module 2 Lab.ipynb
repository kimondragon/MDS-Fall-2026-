```python
import time
import numpy as np


def store_with_np_append(n):
    """
    Store n sensor readings by repeatedly appending to a NumPy array.

    Parameters
    ----------
    n : int
        Number of sensor readings to generate.

    Returns
    -------
    numpy.ndarray
        NumPy array containing readings from 0 to n - 1.

    """
    readings = np.array([])

    for i in range(n):
        readings = np.append(readings, i)

    return readings


def store_with_list(n):
    """
    Store n sensor readings in a Python list, then convert to NumPy once.

    Parameters
    ----------
    n : int
        Number of sensor readings to generate.

    Returns
    -------
    numpy.ndarray
        NumPy array containing readings from 0 to n - 1.


    """
    readings = []

    for i in range(n):
        readings.append(i)

    return np.array(readings)


def seconds(task, n):
    """
    Measure how long task(n) takes to run.

    The function runs the task twice and returns the faster time to reduce
    the effect of temporary system activity on the measurement.

    Parameters
    ----------
    task : function
        Function to benchmark.
    n : int
        Input size passed to the function.

    Returns
    -------
    float
        Best execution time, in seconds, from two runs.
    """
    best = float("inf")

    for _ in range(2):
        start = time.perf_counter()

        task(n)

        elapsed = time.perf_counter() - start
        best = min(best, elapsed)

    return best


# Test each method using increasingly large input sizes.
sizes = [20_000, 40_000, 80_000]

print(
    f"{'method':<21} "
    f"{'n=20,000':>10} "
    f"{'n=40,000':>10} "
    f"{'n=80,000':>10}   "
    f"ratios"
)

for task in [store_with_np_append, store_with_list]:

    # Measure the execution time for each input size.
    t1 = seconds(task, sizes[0])
    t2 = seconds(task, sizes[1])
    t3 = seconds(task, sizes[2])

    # Compare how much the runtime changes when n doubles.
    ratio_1 = t2 / t1
    ratio_2 = t3 / t2

    print(
        f"{task.__name__:<21} "
        f"{t1:9.4f}s "
        f"{t2:9.4f}s "
        f"{t3:9.4f}s   "
        f"{ratio_1:.1f}x, {ratio_2:.1f}x"
    )
```

```python
import random

def make_heart_rates():
    """
    Generate one simulated heart rate reading per minute for a 24-hour day.

    The dataset contains 1,440 readings, corresponding to:
        24 hours × 60 minutes = 1,440 minutes

    Most readings represent a resting heart rate between 58 and 80 bpm.

    Returns
    -------
    list
        A list of 1,440 integer heart rate readings in beats per minute (bpm).
        The index of each value represents the minute of the day:
            index 0    -> 12:00 AM
            index 60   -> 1:00 AM
            index 1025 -> 5:05 PM
    """

    # Create a local random-number generator with a fixed seed.
    rng = random.Random(5012)
    rates = []
    for minute in range(1440):
        rate = rng.randint(58, 80)
        # Simulate a workout from 5:05 PM through 5:49 PM.
        if 17 * 60 + 5 <= minute < 17 * 60 + 50:
            rate = rate + (minute - (17 * 60 + 5)) * 2
        rates.append(rate)

    return rates
```

```python
def linear_search(data, target):
    """
    Search for the first occurrence of a target value using linear search.
    Linear search examines the elements one at a time, starting from the
    beginning of the sequence. The search stops as soon as the target is found.

    Parameters
    ----------
    data : list
        The sequence of values to search.

    target :
        The value to search for.

    Returns
    -------
    int
        The index of the first occurrence of target if it is found.
        Returns -1 if target does not appear in data.
    """

    
    for i in range(len(data)):

        # Check whether the current value matches the target.
        if data[i] == target:

            # Stop immediately and return the index of the first match.
            return i

    return -1
```

```python
import random
import time
import pandas as pd


def make_transactions(n):
    """n transactions (id, time in seconds after midnight, amount),
    almost in time order."""
    
    rng = random.Random(n)

    # Generate transaction times as seconds after midnight.
    times = sorted(
        rng.choices(
            range(8 * 3600, 20 * 3600),
            k=n
        )
    )

    # About 2% of transactions arrive a few positions out of order.
    for _ in range(n // 50):
        i = rng.randrange(n - 5)
        j = i + rng.randint(1, 5)
        times[i], times[j] = times[j], times[i]

    return [
        (
            f"T{k:06d}",          # transaction ID
            t,                    # time in seconds after midnight
            rng.randint(1, 500)   # transaction amount
        )
        for k, t in enumerate(times)
    ]


def sort_by_time(txns):
    """Step 1: put the transactions in time order."""
    
    txns = list(txns)

    for i in range(len(txns) - 1):
        earliest = i

        for j in range(i + 1, len(txns)):
            if txns[j][1] < txns[earliest][1]:
                earliest = j

        txns[i], txns[earliest] = txns[earliest], txns[i]

    return txns


def review_queue(txns):
    """Step 2: largest amounts first;
    equal amounts must stay in time order."""

    table = pd.DataFrame(
        txns,
        columns=["id", "time", "amount"]
    )

    # Convert seconds after midnight to a readable clock time
    table["Clock Time"] = pd.to_datetime(
        table["time"],
        unit="s"
    ).dt.strftime("%-I:%M:%S %p")

    return table.sort_values(
        "amount",
        ascending=False
    )
```

```python
# Measure how long Step 1 takes as the dataset grows.
# We test two input sizes so we can compare how the runtime changes
# when the number of transactions doubles.
for n in [2500 , 5000,10000,200000]:
    txns = make_transactions(n)
    start = time.perf_counter()
    by_time = sort_by_time(txns)
    print(f"n = {n:,}: step 1 took {time.perf_counter() - start:.3f} s")

# Inspect all transactions with amount == 500.
# These rows have the same amount, so check whether they remain in chronological order.
queue = review_queue(by_time)
print(queue[queue["amount"] == 500])   
```

These records are stored in an internal experiment tracking system. Data scientists use the system throughout the day to retrieve previous model runs, inspect model performance, compare experiments, and reproduce earlier results.

The system has the following workload:
- each morning, about 2 million records are loaded, in no particular order
- during the day, users make thousands of lookups by run ID, such as "find run M382910."
- about 20 new runs are added during the day
- run IDs are random, not sequential and each new run must be findable immediately
- the server has very little spare memory
- every run ID is unique

- Data structure: dynamic array
  - Why: 
      - each morning, about 2 million records are loaded (new records are being add meaning that size is increasing)
- Search algorithm: binary search
  - Why: 
      - during the day, users make thousands of lookups by run ID (there is a large quantity of people searching in which a fast seach is required)
- Should the data be sorted before searching? Yes or no and why
  - Yes because binary search needss the data to be sorted.
- Sorting algorithm: insertion, selection, bubble, merge, quick or none
   - Why: quick search
      - the server has very little spare memory (it has extra space for memory)
- Time complexity of adding new records: O(n)
- Time complexity of sorting: O(nlogn)
- Time complexity of searching: O(logn)

   - Why:
      - the bank wants sorting time to be predictable (This allow there to be a predictable case)

**Scenario 3 [5 points]**

An environmental data science team collects temperature measurements from a network of sensors. Each sensor observation contains a timestamp and the recorded temperature.

For example:

readings = [
    ("10:00", 21.4),
    ("10:01", 21.7),
    ("10:02", 21.8),
    ("10:04", 22.1),
    ("10:05", 22.3)
]

late_reading = ("10:03", 21.9)  # delayed by the network; arrives after "10:05"

In most cases, observations reach the server shortly after they are recorded. Occasionally, however, network delays cause a reading to arrive later than expected.

The data collection system behaves as follows:

- the server doesn't know in advance how many readings will arrive
- most readings arrive in time order, but a few are delayed and arrive after later readings
- after the day's readings are collected, analysts frequently search for readings by timestamp


- Data structure: dynamic array
  - Why: 
    - the server doesn't know in advance how many readings will arrive (This mean there has to be avability to add large or small new readings)
- Search algorithm: binary search
  - Why: 
    - analysts frequently search for readings by timestamp (There is a timestamp, so a ninary search would be faster)
- Should the data be sorted before searching? 
  - Yes because the data is already being sorted by timestamps allowing it to be sorted before searching.
- Sorting algorithm: insertion
   - Why: 
      - analysts frequently search for readings by timestamp (This measn that the data is already in order allowing insertion to work best)
- Time complexity of adding new records: O(n)
- Time complexity of sorting: O(n^2)
- Time complexity of searching: O(logn)

**Scenario 4 [5 points]**

A city transportation team collects data from a bike share system. Each completed trip creates a record containing the trip ID, the time the ride started, and its duration.

For example: (trip_id, start_time, duration)

Data scientists later use these records to study travel patterns, identify high demand periods, analyze trip durations, and understand how people use the bike share network.

The system records trips as follows:

- about 500,000 trips are recorded each day, but the exact number isn't known in advance
- a trip is recorded when it ends
- trip IDs are assigned in increasing order as trips are recorded: T000001, T000002, …
- during the day, analysts often look up a trip by trip_id
- at the end of the day, the trips must be put in order of start_time for the next day's analyses
- because trips are recorded when they end, a long ride that started at 9:00 may be recorded after short rides that started at 10:30
- trips with the same start_time must stay in the order they were recorded
- Memory is not a concern overnight

The same dataset is therefore used in different ways: individual trips are retrieved by their IDs, while later analysis requires the records to be organized by start time.

- Data structure: dynamic array
  - Why: 
      - about 500,000 trips are recorded each day, but the exact number isn't known in advance (There is an large increase of records daily bu there isn't an exact number, which dynamic array allows continous growth)
- Search algorithm: binary search
  - Why: 
      - trip IDs are assigned in increasing order as trips are recorded (This means that the data is already sorted, which allows a quick binary search)
- Should the data be sorted before searching? 
    - No, the data has already been sorted in increasing order.
- Sorting algorithm: merge
   - Why: 
    - trips with the same start_time must stay in the order they were recorded (This means that there's a certain order already placed)
      
- Time complexity of adding new records: O(1)
- Time complexity of sorting: O(nlogn)
- Time complexity of searching: O(logn)

The device has the following characteristics:

  - Why:

  - Why:

- Should the data be sorted before searching?
  - No because searches are few and linear search can search through unsorted data
- Sorting algorithm: selection
   - Why: 
      - comparing two values is cheap, but writing or swapping values is expensive (This means that the less swaps the more money is saved)
- Time complexity of adding new records: n/a
- Time complexity of sorting: O(n^2)

  - Why:

    - a monitor scans the queue for jobs with status "corrupt" and removes them from wherever they are (The data is not sorted in a certain form, so it is just a process of going through every single one)
- Should the data be sorted before searching? 
    - No because there is no format for sorting so it's just processed on how it comes in
- Sorting algorithm: none
   - Why:
      - No sorting is neccesary, since it is process as it come in
- Time complexity of adding new records: O(1)
- Time complexity of sorting: n/a
