### Question 5 [10 points]

````python
def weighted_activity_score(values):

    n = len(values) #c1*1

    total = 0 #c2*1
    operations = 0 #c3*1

    for i in range(n): #c4*n

        weight = 0 #c5*n

        for j in range(5): #6c*5n

            weight += j #c7*5n
            operations += 1 #c8*5n

        total += values[i] * weight #c9*n

        if values[i] > 100: #c10*n
            total += 10 #c11*n
        else:
            total += 1

    average = total / n #c12

    return total, average, operations #c13
````

````python
def compare_observations(values, threshold):

    n = len(values) #c1

    comparisons = 0 #c2
    close_pairs = 0 #c3
    total_difference = 0 #c4

    for i in range(n): #c5*n

        for j in range(n):#c6*n^2

            difference = values[i] - values[j] #c7*n^2

            if difference < 0: #c8*n^2
                difference = -difference #c9*n^2

            total_difference += difference #c10*n^2
            comparisons += 1 #c11*n^2

            if difference <= threshold: #c12*n^2
                close_pairs += 1 #c13*n^2

    return comparisons, close_pairs, total_difference #c14
````

````python
def repeated_record_processing(values):

    n = len(values) #c1

    remaining = n #c2

    total = 0 #c3
    operations = 0 #c4

    while remaining > 1: #c5(log2n+1)

        for i in range(n): #c6(nlog2n)

            if values[i] > 0: #c7(lnog2n)
                total += values[i] #c8(nlog2n)
            else:
                total += 1

            operations += 1 #c9(nlog2n)

        remaining = remaining // 2 #c10(log2n)

    return total, operations #c11
````
